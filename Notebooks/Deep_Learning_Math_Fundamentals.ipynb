# Linear Functions

---


A linear function is the simplest type of mathematical relationship between input and output. It describes a **straight line** (in 1D) or a **flat plane** (in higher dimensions).

In one dimension (1D), the equation of a linear function is:

$$
y = \beta + \omega x
$$

- $\beta$ is called the **bias** or **intercept** — it moves the line up or down.  
- $\omega$ is called the **slope** or **weight** — it tilts the line.

In two dimensions (2D), the linear function generalizes to:

$$
y = \beta + \omega_1 x_1 + \omega_2 x_2
$$

- $x_1$ and $x_2$ are two input features.  
- $\omega_1$ and $\omega_2$ control how each input affects the output.


 Any function that is **not in this form** (e.g., involves $x^2$, $\sin(x)$, etc.) is considered **non-linear**.


```python
import numpy as np
import matplotlib.pyplot as plt

def Linear_Function_1D(x, beta, omega):
    return beta + omega * x

x = np.arange(0, 10, 0.01)

beta = 3.0
omega = 1.5

y = Linear_Function_1D(x, beta, omega)


plt.figure(figsize=(7, 4))
plt.plot(x, y, label=f"beta={beta}, omega={omega}")

plt.axhline(0, color='black', lw=0.5, ls='--')
plt.xlabel('x')
plt.ylabel('y')
plt.title('1D Linear Function')
plt.legend()
plt.grid(True)
plt.show()
```

# Linear Function in Two Dimensions

In two dimensions, we have two input variables: $x_1$ and $x_2$.  
The linear function becomes:

$$
y = \beta + \omega_1 x_1 + \omega_2 x_2
$$

- $\omega_1$ controls the effect of $x_1$  
- $\omega_2$ controls the effect of $x_2$  
- $\beta$ still shifts the output up or down

---

Instead of a **straight line** (as in 1D), in 2D this equation represents a **plane**!

```python
from mpl_toolkits.mplot3d import Axes3D

beta = 3.0
omega1 = 1.0
omega2 = 2.0

x1_vals = np.linspace(-5, 5, 50)
x2_vals = np.linspace(-5, 5, 50)
X1, X2 = np.meshgrid(x1_vals, x2_vals)


Y = beta + omega1 * X1 + omega2 * X2

fig = plt.figure(figsize=(7, 4))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X1, X2, Y, cmap='plasma', edgecolor='k', alpha=0.85)


ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$y$')
ax.set_title('3D Linear Plane: $y = 3 + 1x_1 + 2x_2$')

plt.tight_layout()
plt.show()
```



Suppose we have **two outputs** ($y_1$, $y_2$) and **three inputs** ($x_1$, $x_2$, $x_3$):

$$
\begin{aligned}
y_1 &= \beta_1 + \omega_{11}x_1 + \omega_{12}x_2 + \omega_{13}x_3 \\
y_2 &= \beta_2 + \omega_{21}x_1 + \omega_{22}x_2 + \omega_{23}x_3
\end{aligned}
$$

We can write this in **matrix form** as:

$$
\mathbf{y} = \boldsymbol{\beta} + \boldsymbol{\Omega} \mathbf{x}
$$


Where:

- **Input vector** $\mathbf{x}$:
  $$
  \mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix}
  $$

- **Weight matrix** $\boldsymbol{\Omega}$:
  $$
  \boldsymbol{\Omega} =
  \begin{bmatrix}
  \omega_{11} & \omega_{12} & \omega_{13} \\
  \omega_{21} & \omega_{22} & \omega_{23}
  \end{bmatrix}
  $$

- **Bias vector** $\boldsymbol{\beta}$:
  $$
  \boldsymbol{\beta} = \begin{bmatrix} \beta_1 \\ \beta_2 \end{bmatrix}
  $$

- **Output vector** $\boldsymbol{y}$:
  $$
  \mathbf{y} = \begin{bmatrix} y_1 \\ y_2 \end{bmatrix}
  $$

This form is compact, scalable, and ideal for implementation in machine learning libraries like NumPy, PyTorch, or TensorFlow.


```python
def linear_function_3D(x1, x2, x3, beta, omega1, omega2, omega3):
    return beta + omega1 * x1 + omega2 * x2 + omega3 * x3

x1 = 4
x2 = -1
x3 = 2

beta1 = 0.5
beta2 = 0.2
omega11, omega12, omega13 = -1.0, 0.4, -0.3
omega21, omega22, omega23 = 0.1, 0.1, 1.2

y1 = linear_function_3D(x1, x2, x3, beta1, omega11, omega12, omega13)
y2 = linear_function_3D(x1, x2, x3, beta2, omega21, omega22, omega23)

x_vec = np.array([[x1], [x2], [x3]])

omega_mat = np.array([
    [omega11, omega12, omega13],
    [omega21, omega22, omega23]
])

beta_vec = np.array([[beta1], [beta2]])

y_vec = beta_vec + np.matmul(omega_mat, x_vec)

print("Individual Equations:")
print(f"y1 = {beta1} + ({omega11} * {x1}) + ({omega12} * {x2}) + ({omega13} * {x3}) = {y1:.3f}")
print(f"y2 = {beta2} + ({omega21} * {x1}) + ({omega22} * {x2}) + ({omega23} * {x3}) = {y2:.3f}")

print("\nMatrix-Vector Form:")

print("x vector:")
print(x_vec)

print("\nOmega (weight) matrix:")
print(omega_mat)

print("\nBeta (bias) vector:")
print(beta_vec)

print("\nComputed output vector (y = β + Ωx):")
print(y_vec)
```

# Exponential Function

**The exponential function** is the unique real function that maps zero to one and has a derivative everywhere equal to its value.

### Notation and Definition

* **Notation:** It is denoted as $e^x$ or $\exp(x)$.
* **Preference:** The $\exp(x)$ notation is preferred when the argument $x$ is a complicated expression.
* **Origin of Name:** It is called "exponential" because its argument acts as an exponent to which a constant number $e$ ($e \approx 2.718$), the base, is raised.


```python
x_exp = np.linspace(-3, 3, 400)
x_log = np.linspace(0.01, 10, 400)

y_exp = np.exp(x_exp)
y_log = np.log(x_log)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Plot exponential
axs[0].plot(x_exp, y_exp, color='green')
axs[0].set_title('Exponential Function: $y = \exp(x)$')
axs[0].set_xlabel('$x$')
axs[0].set_ylabel('$\exp(x)$')
axs[0].grid(True)

# Plot logarithm
axs[1].plot(x_log, y_log, color='purple')
axs[1].set_title('Logarithm Function: $y = \log(x)$')
axs[1].set_xlabel('$x$')
axs[1].set_ylabel('$\log(x)$')
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

```python


# 1. Create 1D arrays
x_exp = np.linspace(-3, 3, 100)
y_dim = np.linspace(-3, 3, 100) # New Y dimension

x_log = np.linspace(0.01, 10, 100) # avoid log(0)
y_log_dim = np.linspace(-3, 3, 100)

# 2. Create 2D coordinate grids (Meshgrids)
X_exp, Y_exp = np.meshgrid(x_exp, y_dim)
X_log, Y_log = np.meshgrid(x_log, y_log_dim)

# 3. Compute Z values (We apply the math to the X grid)
Z_exp = np.exp(X_exp)
Z_log = np.log(X_log)

# 4. Create 3D subplots
fig = plt.figure(figsize=(14, 6))

# Plot 3D Exponential
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.plot_surface(X_exp, Y_exp, Z_exp, cmap='Greens', alpha=0.8)
ax1.set_title('3D Exponential: $z = \exp(x)$')
ax1.set_xlabel('$x$')
ax1.set_ylabel('$y$ (depth)')
ax1.set_zlabel('$z = \exp(x)$')

# Plot 3D Logarithm
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.plot_surface(X_log, Y_log, Z_log, cmap='Purples', alpha=0.8)
ax2.set_title('3D Logarithm: $z = \log(x)$')
ax2.set_xlabel('$x$')
ax2.set_ylabel('$y$ (depth)')
ax2.set_zlabel('$z = \log(x)$')

plt.tight_layout()
plt.show()
```

```python
# Create x values
# For exp(-x), we can use standard ranges
x_exp = np.linspace(-2, 5, 400)

# For log(x+1), x must be greater than -1 to avoid log(0) and undefined negative roots
x_log = np.linspace(-0.99, 5, 400)

# Compute y values
y_exp = np.exp(-x_exp)
y_log = np.log(x_log + 1)

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Plot exponential decay
axs[0].plot(x_exp, y_exp, color='red', lw=2)
axs[0].set_title('Exponential Decay: $y = \exp(-x)$')
axs[0].set_xlabel('$x$')
axs[0].set_ylabel('$\exp(-x)$')
axs[0].grid(True, alpha=0.5)
axs[0].axhline(0, color='black', lw=0.5)
axs[0].axvline(0, color='black', lw=0.5)

# Plot shifted logarithm
axs[1].plot(x_log, y_log, color='blue', lw=2)
axs[1].set_title('Shifted Logarithm: $y = \log(x+1)$')
axs[1].set_xlabel('$x$')
axs[1].set_ylabel('$\log(x+1)$')
axs[1].grid(True, alpha=0.5)
axs[1].axhline(0, color='black', lw=0.5)
axs[1].axvline(0, color='black', lw=0.5)

plt.tight_layout()
plt.show()
```

# Connection to Deep Learning:

- **Linear Functions** ➔ form the foundation of every neural network layer.  
- **Vectors / Matrices** ➔ how inputs, weights, and outputs are structured and computed.  
- **Exponentials** ➔ appear in activation functions like **Sigmoid** and **Softmax**.  
- **Logarithms** ➔ used in loss functions like **Cross-Entropy**.
